## Try it: make a small data manifest

A compact manifest helps explain where the example came from and what it can and cannot support.

````python
manifest = {
    "example": "synthetic teaching graph",
    "observed_city_data": False,
    "coordinate_reference_system": None,
    "distance_units": None,
    "random_seed": None,
    "source_data": "constructed in this notebook",
}
for key, value in manifest.items():
    print(f"{key}: {value}")
````

````python
import csv
from io import StringIO

toy_edges = [("A", "B", 120), ("B", "C", 85), ("B", "D", 140)]
buffer = StringIO()
writer = csv.writer(buffer)
writer.writerow(["from_node", "to_node", "length_m"])
writer.writerows(toy_edges)
csv_text = buffer.getvalue()
print(csv_text)
print("Exported segments:", len(toy_edges))
````
